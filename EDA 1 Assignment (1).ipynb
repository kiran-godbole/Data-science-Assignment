```python
sns.boxplot(x = data['ASTV'])
plt.show()
```

```python
import numpy as np

def capping(data, columns):
    for col in columns:
        q1 = data[col].quantile(0.25)
        q3 = data[col].quantile(0.75)
        iqr = q3 - q1

        lower = q1 - (1.5 * iqr)
        upper = q3 + (1.5 * iqr)

        data[col] = np.where(data[col] > upper, upper, 
                             np.where(data[col] < lower, lower, data[col]))
```

```python
#Capping outliers
```

```python
capping(data, data.columns)
```

# 2.	Statistical Summary:

```python
# Central tendency and standard deviation
summary = data.describe().T  # Transpose for better readability
summary['median'] = data.median()
summary['IQR'] = data.quantile(0.75) - data.quantile(0.25)

# Reorder columns for clarity
summary = summary[['mean', 'median', 'std', 'min', '25%', '50%', '75%', 'max', 'IQR']]
print(summary)
```

# 🔍  Key Insights:
* FM, DL, and ALTV have a median of zero, suggesting these features are often inactive or sparse in the data.

* ASTV and ALTV show high standard deviation and interquartile range, indicating strong variability — they could be key predictors.

* Negative minimum values in features like ASTV and MSTV are unusual and may point to data entry or scaling issues.

* UC appears stable and normally distributed, with closely aligned mean and median.
